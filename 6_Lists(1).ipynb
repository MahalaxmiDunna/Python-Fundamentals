```python
student={"name":"allen","age":19, "dept":"cse"}
student["city"]="delhi"
student["age"]=23

print(student)
print(student["age"])
print(student.get("dept"))
```

```python
for i in student.keys():
  print(i)
```

```python
for value in student.values():
  print(value)
```

```python
for key,value in student.items():
  print(key,":",value)
```

```python
dict_square={i:i*i for i in range(1,6)}
print(dict_square)
```

```python
dict_square={i:i*i for i in range(1,6) if i%2==0}
print(dict_square)
```
