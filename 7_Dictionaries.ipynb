```python
#dictionary have key as empid and salary as value,get empid as input and print the respective salary
employees={1:10000,
           2:150000,
           3:35000}
empid=int(input("Enter empid:"))
if empid  in employees:
  print("salary:",employees[empid])
else:
  print("emp id not found")
```

```python
total_sal=sum(employees.values())
num_emp=len(employees)
avg_sal=total_sal/num_emp
print("avarage salary:" ,avg_sal)
```

```python
max_sal=max(employees.values())
print("Maximum_salary",max_sal)
```
