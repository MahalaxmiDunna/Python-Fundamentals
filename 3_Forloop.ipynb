check whether a number is a perfect number using a while loop

if a number is divisible by i number% i=0


```python
number = int(input("enter a number: "))
print(number)
i=1
sum=0
while i<number:
  if number%i==0:
    divisor=i
    sum=sum+divisor
    print(divisor)
  i+=1
if number==sum:
  print(number,"is a perfect number")
else:
  print(number,"is not a perfect number")
```
