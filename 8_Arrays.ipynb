```python
#same data type
#collection of items which is store under a single variable
#ordered, access by using the index
import array
array_int=array.array('i',[1,24,5,6])
print(array_int)
print(array_int[2])
```

```python
#operations in array
#insert
array_int.insert(3,30)
print("After inserting 3, 30",array_int)
array_int.append(79)
print("after appending 79",array_int)
array_int.remove(30)
print("after removing 30",array_int)
array_int.pop()
print("after poping",array_int)
```

```python
#find index of an element
print("Index position of 24 is:",array_int.index(24))
array_int[3]=5
print("After updation",array_int)
print("number of 5 in the array  is:", array_int.count(5))
array_int.reverse()
print("Array after reversing :",array_int)
print("No of elements in the array",len (array_int))
```

```python
#add 5 to each element of array
print("At strat:",array_int)
for i in range(0,len(array_int)):
  array_int[i]=array_int[i]+5
print("after adding:",array_int)
```

```python
#second largest numberin a given array
#[20,20,40,70,60,80]
#largest=80 second=70
c=array.array('i',[20,30,40,50,60,9,65,70])
largest=0
second_largest=0
for i in range(0,len(c)):
  if c[i]>largest:
    second_largest=largest
    largest=c[i]
  elif c[i]>second_largest and c[i]!=largest:
    second_largest=c[i]
  print("the second largest number is:",second_largest)
```
