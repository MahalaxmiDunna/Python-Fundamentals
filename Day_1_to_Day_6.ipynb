```python
#write a programme to calculate electricity bill
#frist 100 units-2rs/per unit
#next  100 units-3rs/per unit
#above 200 units-5rs/per unit
units = int(input("Enter electricity bill units: "))
if units<=100:
  bill=units*2
elif units<=200:
  bill=(100*2)+((units - 100)*3)
else:
  bill=(100*2)+(100*3)+((units-200)*5)
print ("Electricity bill:" , bill)
```
