```python
#count the number of even and odd number in a list.
numbers=[12,3,4,56,676,66,90,3,4,5,4]
even_count=0
odd_count=0
for i in numbers:
  if i%2==0:
    even_count=even_count+1
  else:
    odd_count=odd_count+1
  print(even_count)
  print(odd_count)
```

```python
#check whether a number appears more than 3 times in alist without using count()
n=int(input())
numbers=[10,45,23,89,23,10,23,10,23]
count=0
for i in numbers:
  if i==n:
    count+=1
  if count>3:
    print(n,"appears more than 3 times")
  else:
    print(n,"doesnot appears more than 3 times")
```
