FIBONACCI SERIES


```python
#fibonacci using recursion
#fibonacci series
# 0 1 1 2 3 5 8
def fibonacci(n):
  if n==1:
    return 1
  elif n==0:
    return 0
  else:
    return fibonacci (n-1)+fibonacci(n-2)
  fibonacci(5)
```

```python
#factorial 3!=3*2*1= 5!=5*4*3*2*1=120
n=int(input("Enter number:"))
def factorial(n):
  fact=1
  for i in range (n,1,-1):
    fact=fact*i
  print(fact)
factorial(n)
```

```python
def factorial(n):
  if n==0:
    return 1
  else:
    return n * factorial(n-1)

print(factorial(5))
```
